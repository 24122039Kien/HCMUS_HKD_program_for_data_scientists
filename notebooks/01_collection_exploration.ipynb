##### a) Chuẩn hóa định dạng tên cột và ép kiểu dữ liệu GDP
> **Vấn đề chất lượng dữ liệu:**
> - Cột GDP gốc trong file CSV có khoảng trắng thừa: `' gdp_for_year ($) '`.
> - Cột `gdp_for_year` bị lưu dưới dạng chuỗi (`object`) do chứa dấu phẩy phân cách hàng nghìn (ví dụ: `"2,156,624,900"`).
>
> **Giải pháp:** Cắt khoảng trắng, ép kiểu sang `int64`, đồng thời tạo alias tên chuẩn hóa để tương thích với các phân tích tiếp theo.

```python
# --- Chuẩn hóa tên cột và ép kiểu gdp_for_year ---
# Tạo các cột chuẩn hóa không dấu cách/ký tự đặc biệt
if 'gdp_for_year' not in df.columns:
    raw_gdp_col = [c for c in df.columns if 'gdp_for_year' in c][0]
    df['gdp_for_year'] = (
        df[raw_gdp_col]
        .astype(str)
        .str.replace(',', '', regex=False)
        .str.strip()
        .astype('int64')
    )
    df['gdp_per_capita'] = df['gdp_per_capita ($)']
    df['suicides_per_100k'] = df['suicides/100k pop']

print("Đã ép kiểu gdp_for_year thành công sang int64!")
print(f"Kiểu dữ liệu gdp_for_year: {df['gdp_for_year'].dtype}")
print(f"3 giá trị mẫu: {df['gdp_for_year'].head(3).tolist()}")
```

##### b) Tổng quan thống kê mô tả 5 cột số (Overview)
Khảo sát các đại lượng thống kê cơ bản: giá trị trung bình, trung vị (50%), độ lệch chuẩn, min, max và tỷ lệ khuyết thiếu.

##### c) Kiểm tra tính hợp lý (Sanity Checks) & Sai số làm tròn
Kiểm tra các vi phạm logic nghiệp vụ y tế và kinh tế:
1. Có ca tự tử âm (`suicides_no < 0`) không?
2. Có dân số bằng 0 (`population == 0`) hoặc GDP âm không?
3. Sai số làm tròn giữa cột `suicides/100k pop` có sẵn và công thức tính lại từ số ca và dân số.
4. Tính nhất quán của GDP giữa 12 dòng con trong cùng một `country-year`.

```python
print("=" * 60)
print("SANITY CHECKS")
print("=" * 60)

# 3.1a: Số ca tự tử âm (không hợp lệ về mặt vật lý)
neg_suicides = df[df['suicides_no'] < 0]
print(f"\n[1] Số ca tự tử âm (suicides_no < 0):")
print(f"    Số dòng bị ảnh hưởng: {len(neg_suicides):,}")
if len(neg_suicides) > 0:
    display(neg_suicides[['country', 'year', 'sex', 'age', 'suicides_no']].head())
else:
    print("    ✅ Không có giá trị âm.")

# 3.1b: Dân số bằng 0 (gây lỗi phép chia khi tính tỷ lệ)
zero_pop = df[df['population'] == 0]
print(f"\n[2] Dân số bằng 0 (population == 0):")
print(f"    Số dòng bị ảnh hưởng: {len(zero_pop):,}")
if len(zero_pop) > 0:
    display(zero_pop[['country', 'year', 'sex', 'age', 'population']].head())
else:
    print("    ✅ Không có dân số bằng 0.")

# 3.1c: Số ca tự tử = 0 nhưng tỷ lệ > 0 (mâu thuẫn logic)
conflict = df[(df['suicides_no'] == 0) & (df['suicides_per_100k'] > 0)]
print(f"\n[3] Mâu thuẫn: suicides_no=0 nhưng suicides_per_100k > 0:")
print(f"    Số dòng bị ảnh hưởng: {len(conflict):,}")
if len(conflict) > 0:
    display(conflict[['country', 'year', 'sex', 'age', 'suicides_no', 'suicides_per_100k']].head())
else:
    print("    ✅ Không có mâu thuẫn logic.")

# 3.1d: GDP âm (không hợp lệ)
neg_gdp = df[df['gdp_per_capita'] <= 0]
print(f"\n[4] GDP bình quân đầu người <= 0:")
print(f"    Số dòng bị ảnh hưởng: {len(neg_gdp):,}")
if len(neg_gdp) > 0:
    display(neg_gdp[['country', 'year', 'gdp_per_capita']].drop_duplicates().head())
else:
    print("    ✅ Không có GDP <= 0.")
```

### 3.2 Kiểm tra sai số làm tròn của cột `suicides_per_100k`

Cột `suicides_per_100k` trong bộ dữ liệu gốc được tính sẵn, nhưng có thể bị sai số làm tròn.  
Kiểm tra bằng cách tính lại theo công thức chuẩn: $\text{Rate}_{\text{calc}} = 100{,}000 \times \frac{\text{suicides\_no}}{\text{population}}$

```python
# Chỉ kiểm tra trên các dòng có population > 0 để tránh phép chia cho 0
df_valid = df[df['population'] > 0].copy()

df_valid['rate_recalculated'] = 100_000 * df_valid['suicides_no'] / df_valid['population']
df_valid['rate_error'] = df_valid['suicides_per_100k'] - df_valid['rate_recalculated']
df_valid['rate_error_abs'] = df_valid['rate_error'].abs()

print("=" * 60)
print("KIỂM TRA SAI SỐ LÀM TRÒN: suicides_per_100k vs tính lại")
print("=" * 60)
print(f"Số dòng có population > 0 : {len(df_valid):,}")
print(f"Sai số tuyệt đối tối đa   : {df_valid['rate_error_abs'].max():.6f}")
print(f"Sai số tuyệt đối trung bình: {df_valid['rate_error_abs'].mean():.6f}")
print(f"Số dòng sai số > 0.01     : {(df_valid['rate_error_abs'] > 0.01).sum():,}")
print(f"Số dòng sai số > 0.5      : {(df_valid['rate_error_abs'] > 0.5).sum():,}")

# Top 5 dòng sai số lớn nhất
print("\nTop 5 dòng có sai số lớn nhất:")
display(
    df_valid.nlargest(5, 'rate_error_abs')[
        ['country', 'year', 'sex', 'age', 'suicides_no', 'population',
         'suicides_per_100k', 'rate_recalculated', 'rate_error']
    ]
)
```

##### e) Phân tích ngoại lai (Outliers) từ quốc gia dân số siêu nhỏ
> **Bẫy thống kê mẫu số nhỏ (Small denominator problem):**
> Tại các quốc gia có dân số rất ít (như San Marino, Grenada, Anguilla), chỉ cần 1–2 ca tử vong xảy ra ngẫu nhiên cũng có thể đẩy tỷ lệ trên 100,000 dân lên mức cực cao (100–200 ca/100k). Đây là outlier tự nhiên do kích thước mẫu, không phải lỗi nhập liệu.

```python
# Gộp theo country-year để xem tổng dân số và tổng ca của từng quốc gia
df_country_year = (
    df.groupby(['country', 'year'])
    .agg(
        total_suicides = ('suicides_no', 'sum'),
        total_population = ('population', 'sum'),
        gdp_per_capita = ('gdp_per_capita', 'first')
    )
    .reset_index()
)
df_country_year['rate_per_100k'] = (
    100_000 * df_country_year['total_suicides'] / df_country_year['total_population']
)

# Tổng dân số trung bình qua các năm của từng quốc gia
country_avg_pop = (
    df_country_year.groupby('country')['total_population']
    .mean()
    .sort_values()
)

# Xác định IQR-based outlier (cả hai phía)
Q1, Q3 = country_avg_pop.quantile(0.25), country_avg_pop.quantile(0.75)
IQR = Q3 - Q1
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print("=" * 60)
print("PHÂN TÍCH OUTLIER: QUY MÔ DÂN SỐ QUỐC GIA")
print("=" * 60)
print(f"Q1 (25%) = {Q1:,.0f} | Q3 (75%) = {Q3:,.0f} | IQR = {IQR:,.0f}")
print(f"Ngưỡng thấp (IQR) = {lower_fence:,.0f}")
print(f"Ngưỡng cao (IQR) = {upper_fence:,.0f}")

print("\n10 QUỐC GIA DÂN SỐ NHỎ NHẤT (outlier phía dưới):")
small_countries = country_avg_pop.head(10)
for c, v in small_countries.items():
    print(f"  {c:<30} avg_pop = {v:>15,.0f}")

print("\n10 QUỐC GIA DÂN SỐ LỚN NHẤT (outlier phía trên):")
large_countries = country_avg_pop.tail(10)[::-1]
for c, v in large_countries.items():
    print(f"  {c:<30} avg_pop = {v:>15,.0f}")
```

```python
# Minh họa: Tỷ lệ / 100k theo quy mô dân số (xác nhận outlier từ quốc gia siêu nhỏ)
fig, ax = plt.subplots(figsize=(10, 6))

# Scatter plot: dân số vs tỷ lệ / 100k
ax.scatter(
    df_country_year['total_population'],
    df_country_year['rate_per_100k'],
    alpha=0.25, s=15, color='steelblue', label='Các quan sát (country-year)'
)

# Highlight quốc gia siêu nhỏ (dân số < 100,000)
tiny = df_country_year[df_country_year['total_population'] < 100_000]
ax.scatter(
    tiny['total_population'], tiny['rate_per_100k'],
    color='tomato', s=35, zorder=5,
    label=f'Quốc gia dân số siêu nhỏ (< 100k) — n={len(tiny):,} quan sát'
)

# Annotate một số điểm cực đoan
top_rate = df_country_year[df_country_year['total_population'] < 100_000].nlargest(5, 'rate_per_100k')
for _, row in top_rate.iterrows():
    ax.annotate(
        f"{row['country']}\n({row['year']:.0f})",
        xy=(row['total_population'], row['rate_per_100k']),
        xytext=(10, 5), textcoords='offset points',
        fontsize=7, color='darkred',
        arrowprops=dict(arrowstyle='->', color='darkred', lw=0.8)
    )

ax.set_xscale('log')
ax.set_title(
    'Tỷ lệ Tự tử / 100k Dân vs Quy mô Dân số Quốc gia\n'
    'Phát hiện Outlier từ Quốc gia Dân số Siêu nhỏ\n'
    f'(n = {len(df_country_year):,} quan sát country-year)',
    fontsize=12, fontweight='bold'
)
ax.set_xlabel('Dân số (thang log)', fontsize=11)
ax.set_ylabel('Tỷ lệ Tự tử / 100,000 dân', fontsize=11)
ax.legend(fontsize=9)
ax.xaxis.set_major_formatter(mticker.FuncFormatter(
    lambda x, _: f'{x/1e6:.0f}M' if x >= 1e6 else (f'{x/1e3:.0f}K' if x >= 1e3 else f'{x:.0f}')
))

plt.tight_layout()
save_figure(fig, '../data/processed/tv2_fig2_outlier_population.png')
plt.show()
```

##### f) Ma trận tương quan giữa GDP và Tỷ lệ tử vong do tự tử
Tính toán hệ số tương quan tuyến tính Pearson và tương quan thứ hạng Spearman ở cấp độ quốc gia–năm (`country-year`).

```python
# Dùng bảng đã gộp theo country-year (tránh xem xét tương quan giả do lặp lại dữ liệu)
corr_df = df_country_year[['total_suicides', 'total_population', 'rate_per_100k', 'gdp_per_capita']].copy()
corr_df.columns = ['Tổng ca TT', 'Dân số', 'Tỷ lệ/100k', 'GDP/đầu người']

pearson_corr  = corr_df.corr(method='pearson')
spearman_corr = corr_df.corr(method='spearman')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle(
    'Ma trận Tương quan: GDP và Tỷ lệ Tự tử\n'
    f'(cấp country-year, n = {len(corr_df):,})',
    fontsize=13, fontweight='bold'
)

kw = dict(annot=True, fmt='.3f', cmap='RdBu_r', vmin=-1, vmax=1,
          linewidths=0.5, annot_kws={'size': 11})

sns.heatmap(pearson_corr,  ax=ax1, **kw)
ax1.set_title('Tương quan Pearson (tuyến tính)', fontsize=11, fontweight='bold')

sns.heatmap(spearman_corr, ax=ax2, **kw)
ax2.set_title('Tương quan Spearman (thứ hạng, robust)', fontsize=11, fontweight='bold')

plt.tight_layout()
save_figure(fig, '../data/processed/tv2_fig3_correlation.png')
plt.show()

print("\nNhận xét:")
r_gdp_rate_p = pearson_corr.loc['GDP/đầu người', 'Tỷ lệ/100k']
r_gdp_rate_s = spearman_corr.loc['GDP/đầu người', 'Tỷ lệ/100k']
print(f"  Pearson (GDP/đầu người ↔ Tỷ lệ/100k)  = {r_gdp_rate_p:+.3f}")
print(f"  Spearman (GDP/đầu người ↔ Tỷ lệ/100k) = {r_gdp_rate_s:+.3f}")
print("  → Tương quan ở cấp quốc gia-năm có thể bị nhiễu lớn bởi đặc thù quốc gia.")
print("    Q3 (Giai đoạn 2) sẽ kiểm soát bằng phương pháp sai phân bậc nhất (First-difference).")
```

Đây là thêm một lý do để loại năm 2016, bổ sung cho phát hiện ở mục 3.2.2: nếu giữ lại, tỷ lệ gộp năm 2016 bị đẩy lên giả tạo vì mất đúng nhóm tuổi có tỷ lệ thấp nhất.

```python
# Tỷ lệ gộp = 100.000 x tổng số ca / tổng dân số
# (không lấy trung bình cột suicides/100k pop vì các nhóm có dân số khác nhau)
sa = (df.groupby(['age', 'sex'])[['suicides_no', 'population']]
        .sum().reset_index())
sa['rate'] = 100_000 * sa['suicides_no'] / sa['population']

rate_tbl = (sa.pivot(index='age', columns='sex', values='rate')
              .reindex(age_order))
rate_tbl['male - female'] = rate_tbl['male'] - rate_tbl['female']
rate_tbl['male / female'] = rate_tbl['male'] / rate_tbl['female']
display(rate_tbl.round(2))

# Tỷ trọng số ca so với tỷ trọng dân số của từng nhóm tuổi
by_age = (df.groupby('age')[['suicides_no', 'population']]
            .sum().reindex(age_order))
share = pd.DataFrame({
    '% số ca': 100 * by_age['suicides_no'] / by_age['suicides_no'].sum(),
    '% dân số': 100 * by_age['population'] / by_age['population'].sum(),
    'tỷ lệ /100k': 100_000 * by_age['suicides_no'] / by_age['population'],
    '% dòng có 0 ca': 100 * (df['suicides_no'] == 0).groupby(df['age'])
                          .mean().reindex(age_order),
})
display(share.round(2))

by_sex = df.groupby('sex')[['suicides_no', 'population']].sum()
print("Tỷ lệ gộp theo giới tính (số ca / 100.000 dân):")
print((100_000 * by_sex['suicides_no'] / by_sex['population'])
      .round(2).to_string())
```

##### c) Cột `HDI for year`: mức độ và cơ chế thiếu
HDI là chỉ số cấp quốc gia – năm. Nhóm tách hai nguồn thiếu: (1) cả năm không có HDI; (2) năm có HDI nhưng riêng một số quốc gia bị thiếu. Sau đó so sánh nhóm thiếu và nhóm có HDI ở cấp (`country`, `year`) trong các năm có HDI.

```python
hdi_missing = df['HDI for year'].isna()
print(f"Số dòng thiếu HDI: {hdi_missing.sum():,} / {len(df):,} "
      f"({100 * hdi_missing.mean():.2f}%)")

# HDI là chỉ số cấp quốc gia-năm: 12 dòng của một cặp phải cùng giá trị
hdi_nuniq = df.groupby(['country', 'year'])['HDI for year'].nunique()
print("Số cặp (country, year) có hơn 1 giá trị HDI:", (hdi_nuniq > 1).sum())
print("\nThống kê mô tả HDI (các dòng có giá trị):")
print(df['HDI for year'].describe().round(3).to_string())
```

```python
# Tỷ lệ dòng có HDI theo từng năm
hdi_by_year = 100 * df['HDI for year'].notna().groupby(df['year']).mean()
hdi_years = hdi_by_year[hdi_by_year > 0].index.tolist()
print("Các năm có HDI:", hdi_years)
print("Số dòng thiếu HDI nằm ở các năm không có HDI:",
      f"{hdi_missing[~df['year'].isin(hdi_years)].sum():,}")
print("Số dòng thiếu HDI nằm trong các năm có HDI:",
      f"{hdi_missing[df['year'].isin(hdi_years)].sum():,}")

fig, ax = plt.subplots(figsize=(14, 4.5))
bar_colors = ['#2b5c8f' if v > 0 else '#d9d9d9' for v in hdi_by_year]
ax.bar(hdi_by_year.index, hdi_by_year.values, color=bar_colors)
ax.set_title("Tỷ lệ dòng có giá trị HDI theo năm (1985–2016)")
ax.set_xlabel("Năm quan sát")
ax.set_ylabel("Dòng có HDI (%)")
ax.set_ylim(0, 105)
ax.set_xticks(hdi_by_year.index)
ax.tick_params(axis='x', labelrotation=45)
ax.legend(handles=[plt.Rectangle((0, 0), 1, 1, color='#2b5c8f')],
          labels=[f'Năm có HDI ({len(hdi_years)} năm)'])
plt.tight_layout()
plt.show()
```

```python
# Gộp về cấp (country, year) để xét thiếu HDI trong các năm có HDI
cy = (df.groupby(['country', 'year'])
        .agg(hdi=('HDI for year', 'first'),
             gdp_per_capita=('gdp_per_capita ($)', 'first'),
             suicides=('suicides_no', 'sum'),
             population=('population', 'sum'))
        .reset_index())
cy['hdi_missing'] = cy['hdi'].isna()
in_hdi = cy[cy['year'].isin(hdi_years)]
print(f"Số cặp (country, year) trong các năm có HDI: {len(in_hdi):,}")

cmp_tbl = (in_hdi.groupby('hdi_missing')
             .agg(n_country_year=('country', 'size'),
                  n_country=('country', 'nunique'),
                  median_gdp_pc=('gdp_per_capita', 'median'),
                  suicides=('suicides', 'sum'),
                  population=('population', 'sum')))
cmp_tbl['rate_100k'] = (100_000 * cmp_tbl['suicides']
                        / cmp_tbl['population'])
display(cmp_tbl.round(2))

# Các quốc gia không có HDI ở bất kỳ năm nào
never = cy.groupby('country')['hdi_missing'].all()
never = never[never].index
never_tbl = (cy[cy['country'].isin(never)].groupby('country')
               [['suicides', 'population']].sum())
never_tbl['rate_100k'] = (100_000 * never_tbl['suicides']
                          / never_tbl['population'])
print(f"\n{len(never)} quốc gia không bao giờ có HDI:")
display(never_tbl.sort_values('population', ascending=False).round(2))

ok = cy.dropna(subset=['hdi'])
r = np.corrcoef(ok['hdi'], np.log(ok['gdp_per_capita']))[0, 1]
print(f"Tương quan Pearson HDI và log(GDP bình quân): r = {r:.3f}")
```

Q5 sẽ định lượng mức sai lệch này theo nhóm tuổi và khu vực địa lý.

##### d) Số mẫu đủ điều kiện cho bài toán Machine Learning Q6
Một mẫu của Q6 là một nhóm (`country`, `sex`, `age`) tại năm $t$, dùng thông tin đến hết năm $t$ để dự báo tỷ lệ năm $t+1$. Mẫu chỉ hợp lệ khi chuỗi có năm $t+1$ liền kề (để có nhãn) và năm $t-1$ liền kề (để tính đặc trưng biến động $\Delta \text{Rate}_{t,t-1}$). Mẫu được gán vào train/test theo năm đích như kế hoạch: train 1990–2010, test 2011–2015.

```python
key = ['country', 'sex', 'age']
panel = df[key + ['year']].sort_values(key + ['year']).copy()
panel['prev_year'] = panel.groupby(key)['year'].shift(1)
panel['next_year'] = panel.groupby(key)['year'].shift(-1)

has_next = panel['next_year'] == panel['year'] + 1
has_prev = panel['prev_year'] == panel['year'] - 1
gap = panel['next_year'].notna() & ~has_next

print(f"Số nhóm (country, sex, age): {panel.groupby(key).ngroups:,}")
print(f"Số cặp (t, t+1) liên tiếp: {has_next.sum():,}")
print(f"Số bộ (t-1, t, t+1) liên tiếp: {(has_next & has_prev).sum():,}")
print(f"Số chỗ chuỗi bị đứt quãng: {gap.sum():,} "
      f"(thuộc {panel.loc[gap, 'country'].nunique()} quốc gia)")
```

```python
# Gán năm đích t+1 vào tập train / test theo kế hoạch Q6
split_bins = [1985, 1989, 2010, 2015, 2016]
split_names = ['loại (trước 1990)', 'train (1990–2010)',
               'test (2011–2015)', 'loại (2016)']
samples = panel[has_next & has_prev].copy()
samples['target_year'] = samples['year'] + 1
samples['split'] = pd.cut(samples['target_year'], bins=split_bins,
                          labels=split_names)

# HDI ở năm đặc trưng t (nếu muốn dùng HDI làm đặc trưng)
samples = samples.merge(cy[['country', 'year', 'hdi']],
                        on=['country', 'year'], how='left')

split_tbl = (samples.groupby('split', observed=True)
               .agg(n_samples=('country', 'size'),
                    n_country=('country', 'nunique'),
                    pct_has_hdi_at_t=('hdi', lambda s: 100 * s.notna()
                                      .mean())))
display(split_tbl.round(1))

train_c = set(samples.loc[samples['split'] == split_names[1], 'country'])
test_c = set(samples.loc[samples['split'] == split_names[2], 'country'])
print("Quốc gia có trong test nhưng không có trong train:",
      sorted(test_c - train_c))
```

```python
# Mức tự tương quan giữa tỷ lệ năm t và t+1 (chỉ dùng giai đoạn train)
rate = df[key + ['year', 'suicides/100k pop']]
nxt = rate.assign(year=rate['year'] - 1).rename(
    columns={'suicides/100k pop': 'rate_next'})
lag_pairs = rate.merge(nxt, on=key + ['year'])
lag_train = lag_pairs[(lag_pairs['year'] + 1).between(1990, 2010)]
print(f"Số cặp (t, t+1) có năm đích 1990–2010: {len(lag_train):,}")
print("Tương quan Pearson giữa rate_t và rate_t+1: "
      f"r = {lag_train['suicides/100k pop'].corr(lag_train['rate_next']):.3f}")

per_year = (samples.groupby(['target_year', 'split'], observed=True)
              .size().reset_index(name='n'))
fig, ax = plt.subplots(figsize=(14, 4.5))
sns.barplot(data=per_year, x='target_year', y='n', hue='split',
            hue_order=split_names, dodge=False, ax=ax)
ax.set_title("Số mẫu (t-1, t, t+1) liên tiếp theo năm đích t+1 cho Q6")
ax.set_xlabel("Năm đích t+1")
ax.set_ylabel("Số mẫu (nhóm country-sex-age)")
ax.tick_params(axis='x', labelrotation=45)
ax.legend(title="Vai trò trong Q6")
plt.tight_layout()
plt.show()
```

> **Nhận xét:**
