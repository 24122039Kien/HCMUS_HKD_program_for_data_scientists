#### 4. Collection method (Cách thức thu thập & Các giới hạn đã biết)
- **Cách thức thu thập:** Dữ liệu hành chính (administrative records) thu thập từ hệ thống đăng ký hộ tịch khai tử (Civil Registration and Vital Statistics - CRVS) và hệ thống giám sát dịch tễ của cơ quan y tế các nước thành viên nộp định kỳ về WHO. Dữ liệu kinh tế được trích xuất từ hệ thống thống kê tài khoản quốc gia của Ngân hàng Thế giới.

- **Khoảng thời gian:** Từ năm 1985 đến 2016 (32 năm).
- **Các giới hạn và thiên lệch đã biết (Known limitations & biases):**
  - *Thiên lệch chọn mẫu (Selection Bias):* Đa số các quốc gia nộp số liệu thuộc khối phát triển (châu Âu, châu Mỹ); các quốc gia nghèo và khu vực châu Phi gần như vắng mặt do hệ thống y tế và đăng ký hộ tịch chưa hoàn thiện.
  - *Báo cáo thiếu (Under-reporting Bias):* Do rào cản tôn giáo, luật pháp và định kiến xã hội, số ca tự tử ở một số nền văn hóa có thể bị ghi nhận nhầm thành tai nạn hoặc nguyên nhân khác.
  - *Thiếu dữ liệu theo thời gian (Temporal Discontinuity):* Không phải quốc gia nào cũng nộp đủ 32 năm liên tục; đặc biệt năm 2016 có rất ít quốc gia hoàn thành báo cáo.
  - *Cột HDI bị khuyết thiếu cao:* Cột `HDI for year` thiếu ~69.6% dữ liệu do UNDP trước năm 2010 chỉ tính toán định kỳ 5 năm một lần.

#### 3.2.2. Khám phá chuyên sâu cột Không gian & Thời gian (`country`, `year`, `country-year`)


Mục tiêu: Đánh giá tính toàn vẹn của hệ thống định danh, độ bao phủ không gian - thời gian, và phát hiện các giới hạn của dữ liệu.

```python
num_countries = df['country'].nunique()
min_year = df['year'].min()
max_year = df['year'].max()
print(f"Tổng số quốc gia ghi nhận: {num_countries} quốc gia")
print(f"Giai đoạn thời gian quan sát: {min_year} đến {max_year} ({max_year - min_year + 1} năm)")

# 1. Kiểm tra tính duy nhất của bộ khóa định danh tổng hợp: (country, year, sex, age)
composite_key_dups = df.duplicated(subset=['country', 'year', 'sex', 'age']).sum()
print(f"Số dòng trùng khóa tổng hợp (country, year, sex, age): {composite_key_dups}")

# 2. Kiểm tra tính đồng nhất giữa cột khóa ghép country-year với 2 cột gốc country và year
expected_country_year = df['country'] + df['year'].astype(str)
mismatch_key = (df['country-year'] != expected_country_year).sum()
print(f"Số dòng không khớp giữa country-year và (country + year): {mismatch_key}")
```

> **Nhận xét về hệ thống khóa:**
> - Khóa định danh tổng hợp `(country, year, sex, age)` hoàn toàn duy nhất (`0` dòng trùng lặp), đảm bảo mỗi nhóm dân số chỉ xuất hiện đúng 1 lần trong mỗi năm.
> - Cột ghép `country-year` hoàn toàn nhất quán với 2 cột thành phần `country` và `year`.

```python
# Thống kê số lượng quốc gia nộp dữ liệu theo từng năm
countries_per_year = df.groupby('year')['country'].nunique().reset_index()
countries_per_year.columns = ['year', 'country_count']

fig, ax = plt.subplots(figsize=(12, 5))
sns.barplot(data=countries_per_year, x='year', y='country_count', color='#2b5c8f', ax=ax)
ax.set_title("Hình 1: Số lượng quốc gia có dữ liệu báo cáo theo từng năm (1985–2016)", fontsize=13, pad=12)
ax.set_xlabel("Năm quan sát (Year)")
ax.set_ylabel("Số lượng quốc gia (Nước)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

> **PHÁT HIỆN QUAN TRỌNG VỀ ĐỘ PHỦ THỜI GIAN (CRITICAL FINDING):**
> 1. **Giai đoạn ổn định (1995–2014):** Số lượng quốc gia báo cáo duy trì tương đối đều đặn trong khoảng từ 65 đến 80 quốc gia/năm.
> 2. **Sụt giảm nghiêm trọng vào năm 2016:** Số lượng quốc gia có dữ liệu năm 2016 **rơi thẳng xuống chỉ còn 16 quốc gia** (thiếu dữ liệu hơn 80% so với các năm trước) do độ trễ trong việc cập nhật báo cáo dịch tễ toàn cầu.
> 3. **Hàm ý thực tiễn (Actionable Insight):** Để tránh sai lệch nghiêm trọng do biến động cỡ mẫu, nhóm quyết định **loại bỏ năm 2016** và giới hạn cửa sổ phân tích chuỗi thời gian chính thức trong giai đoạn **1990 – 2015** (đặc biệt cho câu hỏi Q1).

```python
# Thống kê độ phủ thời gian của từng quốc gia (Quốc gia có dữ liệu liên tục vs ngắt quãng)
years_by_country = df.groupby('country')['year'].nunique().sort_values(ascending=False)

continuous_30plus = (years_by_country >= 30).sum()
between_15_29 = ((years_by_country >= 15) & (years_by_country < 30)).sum()
under_15 = (years_by_country < 15).sum()

print(f"Số quốc gia có dữ liệu liên tục >= 30 năm : {continuous_30plus} quốc gia")
print(f"Số quốc gia có dữ liệu trung bình (15–29 năm): {between_15_29} quốc gia")
print(f"Số quốc gia có dữ liệu ngắn hạn (< 15 năm)    : {under_15} quốc gia")

# Top 10 quốc gia có dữ liệu đầy đủ nhất và ít nhất
summary_coverage = pd.DataFrame({
    'Top 10 quốc gia đầy đủ nhất': years_by_country.head(10).index + ' (' + years_by_country.head(10).astype(str) + ' năm)',
    'Top 10 quốc gia ít dữ liệu nhất': years_by_country.tail(10).index + ' (' + years_by_country.tail(10).astype(str) + ' năm)'
})
summary_coverage
```

> **Nhận xét về tính khả thi của câu hỏi Q1 (Fixed Balanced Cohort):**
> - Có **24 quốc gia có dữ liệu liên tục từ 30 năm trở lên** (như Vương quốc Anh, Hoa Kỳ, Nhật Bản, các nước Bắc Âu...).
> - Điều này chứng minh tính khả thi hoàn toàn để xây dựng một **Tập hợp quốc gia cân bằng cố định (Fixed Balanced Cohort)** cho câu hỏi nghiên cứu **Q1**, nhằm khử hoàn toàn thiên lệch chọn mẫu (Selection Bias) khi so sánh với mẫu nộp ngẫu nhiên hàng năm.
